### Step 3: Data Cleaning

```python
# STEP 1: Getting Important Data from Dataset

# In this DataSet we found that there are more than 2,00,000 rows and 130 columns
# Now, taking only important rows after closely looking the dataset
# These are 17 columns whick is useful for analysis.

imp_columns =[
    'eventid', 'iyear', 'imonth', 'iday', 'country_txt', 'region_txt', 
    'provstate', 'city', 'latitude', 'longitude', 'attacktype1_txt', 
    'targtype1_txt', 'gname', 'weaptype1_txt', 'nkill', 'nwound','success', 'summary'
]

# Copying the data using .copy() that ensures the original data set does not get changed and
# stored in a another variable named as 'data_clean'

data_clean = df1[imp_columns].copy()
```

```python
# STEP 2: Renaming the Column names

# Renaming the name of the columns for easy understanding

rename_col = {
    'iyear': 'Year',
    'imonth': 'Month',
    'iday': 'Day',
    'country_txt': 'Country',
    'region_txt': 'Region',
    'provstate': 'State',
    'city': 'City',
    'latitude': 'Latitude',
    'longitude': 'Longitude',
    'attacktype1_txt': 'Attack_Type',
    'targtype1_txt': 'Target_Type',
    'gname': 'Terrorist_Group',
    'weaptype1_txt': 'Weapon_Type',
    'nkill': 'Killed',
    'nwound': 'Wounded',
    'success': 'Success',
    'summary': 'Summary'
}

data_clean = data_clean.rename(columns=rename_col)
```

```python
# STEP 3: Handle Missing Values

# 3A. Casualties: We logically assume if kills/wounds are empty, they are likely 0.
# We do NOT use the "mean" (average) here, because applying average kills makes fake victims!

data_clean['Killed'] = data_clean['Killed'].fillna(0).astype(int)
data_clean['Wounded'] = data_clean['Wounded'].fillna(0).astype(int)


# Let's combine Killed and Wounded to make a 'Casualties' column.

data_clean['Casualties'] = data_clean['Killed'] + data_clean['Wounded']


# Here we use the 'mean' to fill the Latitude and Longitude

data_clean['Latitude'] = data_clean.groupby('Country')['Latitude'].transform(lambda x: x.fillna(x.mean()))
data_clean['Longitude'] = data_clean.groupby('Country')['Longitude'].transform(lambda x: x.fillna(x.mean()))
```

```python
# 3B. Text fields: Fill empty city/state/summaries with 'Unknown'

columns_data = ['State', 'City', 'Summary']
for col in columns_data:
    data_clean[col] = data_clean[col].fillna('Unknown')
```

```python
# 3C : Finding the Duplicates 

print(data_clean.duplicated().sum())
```

```python
# 3D : Here we Triming the WhiteSpace and Case

df = data_clean.map(lambda s:s.upper().strip() if type(s)==str else s)
```

### Dataset columns

* **eventid** – Unique identifier assigned to each recorded incident.
* **Year** – The year in which the event occurred.
* **Month** – The month when the event took place.
* **Day** – The specific day of the event.
* **Country** – The country where the incident happened.
* **Region** – The broader geographical region of the event (e.g., South Asia, Middle East).
* **State** – The state or province within the country.
* **City** – The city where the incident occurred.
* **Latitude** – The geographic coordinate indicating north–south position.
* **Longitude** – The geographic coordinate indicating east–west position.
* **Attack_Type** – The method used in the attack (e.g., bombing, armed assault).
* **Target_Type** – The category of the target (e.g., civilians, military, government).
* **Terrorist_Group** – The group responsible for carrying out the attack.
* **Weapon_Type** – The type of weapon used in the incident.
* **Killed** – The number of people who died in the attack.
* **Wounded** – The number of people injured in the attack.
* **Success** – Indicates whether the attack was successfully carried out (often binary: yes/no).
* **Summary** – A brief textual description of the event.
* **Casualties** – Total number of affected people (usually killed + wounded).

### Step 6: Data Visualization

### Q. What is the year-over-year trend of global terrorist incidents?

```python
# Number of Terrorist Attacks over Years
plt.figure(figsize=(15, 6))
sns.countplot(x='Year', data=df, palette='viridis')
plt.xticks(rotation=90)
plt.title('Number of Terrorist Attacks per Year (1970-2020)')
plt.xlabel('Year')
plt.ylabel('Number of Attacks')
plt.savefig('attacks_per_year.png')
```

**Inference:** After 2004, there is a continuos raise in Terrorist Attach. Also there is a highest number of Terrorist Attack in 2014.

### Q. Where is the highest concentration of global attacks occurring?

```python
# Top 10 Countries with Most Attacks
plt.figure(figsize=(10, 5))
top_countries = df['Country'].value_counts().head(10)
sns.barplot(x=top_countries.index, y=top_countries.values, palette='viridis')
plt.title('Top 10 Countries Affected by Terrorism')
plt.xlabel('Country')
plt.ylabel('Number of Attacks')
plt.xticks(rotation=45);
# plt.savefig('top_countries.png')
```

**Inference:** Iraq faces the most Terrorist Attack. Followed by Afghanistan, Pakistan, India are affected by Terrorist Attack.

### Q. How are different attack methods distributed?

```python
# Most Common Attack Types Distribution
attack_counts = df['Attack_Type'].value_counts()
top6   = attack_counts.head(6)
other  = attack_counts.iloc[6:].sum()
labels = list(top6.index) + ['Other']
sizes  = list(top6.values) + [other]

plt.figure(figsize=(9, 9))
plt.pie(
    sizes, labels=labels, autopct='%1.1f%%', startangle=140,
    pctdistance=0.82
)

plt.title('Attack Type Distribution', fontsize=14, fontweight='bold')
plt.show();
# plt.savefig('attack_types.png')
```

**Inference:** Bombings and explosions are the primary method of attack, used in 47% of cases nearly double the rate of armed assaults (24%).  
Bombing and explosive cause large damage than the Armed Assault.  
Here we can see Terrorist attack from distance using Explosives because there is lower risk but high impact.

### Q. How does the distribution of different attack types vary across regions?

```python
# Attack Type v/s Region

cross = pd.crosstab(df['Region'], df['Attack_Type'])

cross.plot(kind='bar', stacked=True, figsize=(11,6))
plt.title("Attack Type vs Region")
plt.show()
```

**Inference:** All of the Regions are mostly affected by Bombing/Explosive followed by the Armed Assault.
